```python
# Import Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

1. Read the image ('Eagle_in_Flight.jpg') using OpenCV imread() as a grayscale image.

```python
# 1. Read the image as a grayscale image
img = cv2.imread('Eagle_in_Flight.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError("Eagle_in_Flight.jpg not found")

plt.imshow(img, cmap='gray')
plt.title("Eagle in Flight - Grayscale")
plt.axis('off')
plt.show()
```

2. Print the image width, height & Channel

```python
# 2. Print image width, height and channel
height, width = img.shape

print("Width:", width)
print("Height:", height)
print("Channel:", 1)
```

3. Display the image using matplotlib imshow().

```python
# 3. Display the image using matplotlib imshow()
plt.imshow(img, cmap='gray')
plt.title("Eagle Image")
plt.axis('off')
plt.show()
```

4. Save the image as a PNG file using OpenCV imwrite().

```python
# 4. Save the image as a PNG file
cv2.imwrite('Eagle_Grayscale.png', img)
print("Image saved as Eagle_Grayscale.png")
```

5. Read the saved image above as a color image using cv2.cvtColor().

```python
# 5. Read the saved image as a color image
img_color = cv2.imread('Eagle_Grayscale.png', cv2.IMREAD_COLOR)

if img_color is None:
    raise FileNotFoundError("Eagle_Grayscale.png not found")

img_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
```

6. Display the Colour image using matplotlib imshow() & Print the image width, height & channel

```python
# 6. Display the color image and print dimensions
plt.imshow(img_color)
plt.title("Eagle Image - Color")
plt.axis('off')
plt.show()

height, width, channel = img_color.shape
print("Width:", width)
print("Height:", height)
print("Channel:", channel)
```

7. Crop the image to extract any specific (Eagle alone)  object from the image.

```python
# 7. Crop the image to extract the eagle
plt.figure(figsize=(10, 6))
plt.imshow(img, cmap='gray')
plt.title("Select Eagle Crop Coordinates")
plt.show()

# Adjust these coordinates if needed
cropped_img = img[100:500, 100:500]

plt.imshow(cropped_img, cmap='gray')
plt.title("Cropped Eagle")
plt.axis('off')
plt.show()
```

8. Resize the image up by a factor of 2x.

```python
# 8. Resize the cropped image by a factor of 2
resized_img = cv2.resize(
    cropped_img, None,
    fx=2, fy=2,
    interpolation=cv2.INTER_CUBIC
)

plt.imshow(resized_img, cmap='gray')
plt.title("Resized Eagle")
plt.axis('off')
plt.show()
```

9. Flip the cropped/resized image horizontally. (Eagle Image)

```python
# 9. Flip the cropped/resized image horizontally
flipped_img = cv2.flip(resized_img, 1)

plt.imshow(flipped_img, cmap='gray')
plt.title("Horizontally Flipped Eagle")
plt.axis('off')
plt.show()
```

10. Read in the image ('Apollo-11-launch.jpg').

```python
# 10. Read Apollo-11-launch.jpg
apollo = cv2.imread('Apollo-11-launch.jpg')

if apollo is None:
    raise FileNotFoundError("Apollo-11-launch.jpg not found")

apollo = cv2.cvtColor(apollo, cv2.COLOR_BGR2RGB)

plt.imshow(apollo)
plt.title("Apollo 11 Launch")
plt.axis('off')
plt.show()
```

11. Add the following text to the dark area at the bottom of the image (centered on the image):

```python
# 11. Add text centered at the bottom
text = 'Apollo 11 Saturn V Launch, July 16, 1969'
font_face = cv2.FONT_HERSHEY_PLAIN
font_scale = 1.5
thickness = 2

apollo_bgr = cv2.cvtColor(apollo.copy(), cv2.COLOR_RGB2BGR)

text_size, baseline = cv2.getTextSize(
    text, font_face, font_scale, thickness
)

text_x = (apollo.shape[1] - text_size[0]) // 2
text_y = apollo.shape[0] - 20

cv2.putText(
    apollo_bgr, text, (text_x, text_y),
    font_face, font_scale, (255, 255, 255),
    thickness, cv2.LINE_AA
)

apollo_annotated = cv2.cvtColor(apollo_bgr, cv2.COLOR_BGR2RGB)
```

12. Draw a magenta rectangle that encompasses the launch tower and the rocket.

```python
# 12. Draw a magenta rectangle around the launch tower and rocket
rect_color = (255, 0, 255)  # Magenta in RGB

# Approximate coordinates; adjust to the actual image
x1 = int(apollo.shape[1] * 0.35)
y1 = int(apollo.shape[0] * 0.10)
x2 = int(apollo.shape[1] * 0.70)
y2 = int(apollo.shape[0] * 0.90)

cv2.rectangle(
    apollo_annotated,
    (x1, y1), (x2, y2),
    rect_color, 3
)
```

13. Display the final annotated image.

```python
# 13. Display the final annotated image
plt.figure(figsize=(12, 8))
plt.imshow(apollo_annotated)
plt.title("Apollo 11 - Annotated Image")
plt.axis('off')
plt.show()

cv2.imwrite(
    'Apollo_11_Annotated.png',
    cv2.cvtColor(apollo_annotated, cv2.COLOR_RGB2BGR)
)
```

14. Read the image ('boy.jpg').

```python
# 14. Read the image boy.jpg
boy_bgr = cv2.imread('boy.jpg')

if boy_bgr is None:
    raise FileNotFoundError("boy.jpg not found")

boy = cv2.cvtColor(boy_bgr, cv2.COLOR_BGR2RGB)

plt.imshow(boy)
plt.title("Original Boy Image")
plt.axis('off')
plt.show()
```

15. Adjust the brightness of the image.

```python
# 15. Create a matrix of ones with float64 data type
matrix_ones = np.ones(boy.shape, dtype=np.float64)

print("Matrix shape:", matrix_ones.shape)
print("Data type:", matrix_ones.dtype)
```

16. Create brighter and darker images.

```python
# 16. Create brighter and darker images
matrix = np.ones(boy.shape, dtype=np.float64) * 50

img_brighter = cv2.convertScaleAbs(boy, alpha=1.0, beta=50)
img_darker = cv2.convertScaleAbs(boy, alpha=1.0, beta=-50)
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
# 17. Display the images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(boy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_darker)
plt.title("Darker Image")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_brighter)
plt.title("Brighter Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

18. Modify the image contrast.

```python
# 18. Create higher contrast images using scale factors 1.1 and 1.2
matrix1 = np.ones(boy.shape, dtype=np.float64) * 1.1
matrix2 = np.ones(boy.shape, dtype=np.float64) * 1.2

img_higher1 = np.clip(
    boy.astype(np.float64) * matrix1, 0, 255
).astype(np.uint8)

img_higher2 = np.clip(
    boy.astype(np.float64) * matrix2, 0, 255
).astype(np.uint8)
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
# 19. Display original, lower contrast and higher contrast images
img_lower = cv2.convertScaleAbs(boy, alpha=0.8, beta=0)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(boy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_lower)
plt.title("Lower Contrast")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_higher1)
plt.title("Higher Contrast (1.1)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
# 20. Split the image into B, G, R components
boy_bgr = cv2.cvtColor(boy, cv2.COLOR_RGB2BGR)
B, G, R = cv2.split(boy_bgr)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(B, cmap='gray')
plt.title("Blue Channel")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(G, cmap='gray')
plt.title("Green Channel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(R, cmap='gray')
plt.title("Red Channel")
plt.axis('off')

plt.tight_layout()
plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
# 21. Merge the R, G, B channels
merged_rgb = cv2.merge((R, G, B))

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(boy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged_rgb)
plt.title("Merged RGB Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
# 22. Split the image into H, S, V components
hsv = cv2.cvtColor(boy, cv2.COLOR_RGB2HSV)
H, S, V = cv2.split(hsv)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(H, cmap='gray')
plt.title("Hue Channel")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(S, cmap='gray')
plt.title("Saturation Channel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(V, cmap='gray')
plt.title("Value Channel")
plt.axis('off')

plt.tight_layout()
plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
# 23. Merge H, S, V and display with original image
merged_hsv = cv2.merge((H, S, V))
merged_hsv_rgb = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2RGB)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(boy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged_hsv_rgb)
plt.title("Merged HSV Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```
